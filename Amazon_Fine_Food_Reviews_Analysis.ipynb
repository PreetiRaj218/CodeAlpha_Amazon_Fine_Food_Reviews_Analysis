```python
import nltk
nltk.download('vader_lexicon')
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()
```

```python
def get_sentiment(text):
    
    score = sia.polarity_scores(str(text))

    if score['compound'] > 0:
        return 'Positive'
    
    elif score['compound'] < 0:
        return 'Negative'
    
    else:
        return 'Neutral'
```

```python
sample_df = df.head(10000).copy()

sample_df['Sentiment'] = sample_df['Text'].apply(get_sentiment)
```

```python
sample_df['Sentiment'].value_counts()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='Sentiment', data=sample_df)

plt.title("Sentiment Analysis of Amazon Reviews")
plt.show()
```
